# Application that allows users to add tasks, view tasks and mark tasks as complete

```python
# Empty list to store tasks
tasks=[]

# Function to add new task
def add_task(new_task):
  tasks.append(new_task)
  print('task added',new_task)

# Function to view all tasks
def view_task():
  if tasks:
    print('tasks: ')
    for index,task in enumerate(tasks,start=1):
      print(index,task)
  else:
    print('No task to display')

# Function to mark a task as complete
def completed_task(task_index):
  if 1<=task_index<=len(tasks):
    completed_task=tasks.pop(task_index-1)
    print('completed task',completed_task)
  else:
    print('Invalid task index')

# Main loop
while True:
  print('Select an option:- ')
  print('1. add tasks')
  print('2. view tasks')
  print('3. complete task')
  print('4. exit')

  choice=input('enter your choice: ')

  if choice=='1':
    new_task=input('enter the task: ')
    add_task(new_task)
  elif choice=='2':
    view_task()
  elif choice=='3':
    task_index=int(input('enter task number to mark as complete: '))
    completed_task(task_index)
  elif choice=='4':
    print('goodbye!')
    break
  else:
    print('invalid choice. please try again')
```
