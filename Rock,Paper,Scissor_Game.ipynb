# Rock Paper Scissor Game in Python

```python
while True:
  import random
  print('rock, scissor, paper')

  print(''' Winning Rules
  1- Rock defeats Scissors.
  2- Scissors defeats Paper.
  3- Paper defeats Rock.
  4- If both players choose the same option, the round ends in a draw. ''')

  user_input=int(input('select:- 1 for rock, 2 for scissor, 3 for paper '))
  if user_input > 3:
    print('invalid input')
  else:
    if user_input==1:
      print('you selected rock')
    elif user_input==2:
      print('you selected scissor')
    elif user_input==3:
      print('you selected paper')

  computer_input=random.randint(1,3)
  if computer_input==1:
    print('computer selected rock')
  elif computer_input==2:
    print('computer selected scissor')
  elif computer_input==3:
    print('computer selected paper')

  if user_input==computer_input:
    print('draw')
  elif user_input==1 and computer_input==2:
    print('you win')
  elif user_input==1 and computer_input==3:
    print('you lose')
  elif user_input==2 and computer_input==1:
    print('you lose')
  elif user_input==2 and computer_input==3:
    print('you win')
  elif user_input==3 and computer_input==1:
    print('you win')
  elif user_input==3 and computer_input==2:
    print('you lose')

  exit=input('do you want to quit y/n : ')
  if exit=='y':
    print('thanks for playing')
    break
```
