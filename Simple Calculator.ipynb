# simple calculator which is used to Add, Subtract, Multiply, and Divide the two numbers.

```python
# Function to add two numbers
def addition(num1,num2):
  return num1+num2

# Function to subtract two numbers
def subtraction(num1,num2):
  return num1-num2

# Function to multiply two numbers
def multiplication(num1,num2):
  return num1*num2

# Function to divide two numbers
def division(num1,num2):
  if num2!=0:
    return num1/num2
  else:
    return 'cannot divide by zero'

# Main loop
while True:
  print('Select operation :- ')
  print('1. for addition')
  print('2. for subtraction')
  print('3. for multiplication')
  print('4. for division')
  print('5. to quit')

  choice=input('enter your choice (1/2/3/4/5): ')

  if choice in ('1','2','3','4'):
    num1=float(input('enter 1st number: '))
    num2=float(input('enter 2nd number: '))

  if choice=='1':
    print('{} + {} ='.format(num1,num2),addition(num1,num2))
  elif choice=='2':
    print('{} - {} ='.format(num1,num2),subtraction(num1,num2))
  elif choice=='3':
    print('{} * {} ='.format(num1,num2),multiplication(num1,num2))
  elif choice=='4':
    print('{} / {} ='.format(num1,num2),division(num1,num2))
  elif choice=='5':
    print('Goodbye!')
    break
  else:
    print('Invalid input')
```
